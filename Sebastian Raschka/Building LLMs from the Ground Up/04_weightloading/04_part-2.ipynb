```python
# 目的：用 LitGPT 的高层 LLM 接口加载 phi-2 并做一次问答生成。
# 对比 part-1：这里下载/加载/生成全被封装，无需手动处理 checkpoint 解析与权重搬运。
from litgpt import LLM

llm = LLM.load("microsoft/phi-2")

llm.generate("What do Llamas eat?")
```

```python
# 目的：演示流式输出 stream=True——逐个 token 产出并即时打印，而非等全部生成完再一次性返回。
# max_new_tokens=200 限制最多新生成 200 个 token。
result = llm.generate("What do Llamas eat?", stream=True, max_new_tokens=200)
for e in result:
    print(e, end="", flush=True)  # end="" 不换行、flush=True 立即刷新，形成“打字机”式实时输出
```

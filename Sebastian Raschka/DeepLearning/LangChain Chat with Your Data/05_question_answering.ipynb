# Question Answering

**RAG 流程小结**：本 notebook 是 RAG 流程最后一步——**问答（Question Answering）**。前面的检索器只负责"找到相关文档"，
这里用 `RetrievalQA` 这个链（Chain）把"检索"和"生成回答"串起来：先用 retriever 找到相关文档，
再把这些文档拼进 prompt 交给 LLM 生成最终答案。同时会对比 stuff / map_reduce / refine 几种不同的"把多个文档喂给 LLM"的策略。

```python
import os
import openai
import sys
sys.path.append('../..')

from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv()) # read local .env file

openai.api_key  = os.environ['OPENAI_API_KEY']
```

```python
# 课程录制时 OpenAI 曾经历一次模型改版（gpt-3.5-turbo-0301 -> gpt-3.5-turbo），
# 这段逻辑是根据当前日期自动选用合适的模型名，避免用到已经过时/下线的旧模型版本号
import datetime
current_date = datetime.datetime.now().date()
if current_date < datetime.date(2023, 9, 2):
    llm_name = "gpt-3.5-turbo-0301"
else:
    llm_name = "gpt-3.5-turbo"
print(llm_name)
```

```python
# 【真实 Bug 修复】重新打开 03/04 节持久化好的 Chroma 向量库。
#
# 原代码 `from langchain_core.vectorstores import Chroma` 是错的：langchain_core.vectorstores
# 这个模块里只有 VectorStore 抽象基类，根本没有 Chroma 这个具体实现，会直接
# `ImportError: cannot import name 'Chroma' from 'langchain_core.vectorstores'`。
# 正确路径应该是 langchain_community.vectorstores（和 03、04 节一致）。
# 另外 `from langchain_community.embeddings.openai import OpenAIEmbeddings` 虽然还能 import（会有 deprecation warning），
# 但官方推荐统一改用独立包 langchain_openai，这里也一并改掉，和其它 notebook 保持一致。
from langchain_community.vectorstores import Chroma
from langchain_openai import OpenAIEmbeddings
persist_directory = 'docs/chroma/'
embedding = OpenAIEmbeddings()
vectordb = Chroma(persist_directory=persist_directory, embedding_function=embedding)
```

```python
# 确认向量库里已经有数据（应该和 03 节存入的块数一致）
print(vectordb._collection.count())
```

```python
# 先单独验证一下检索这一步是否正常工作，再往下接入 LLM 做完整问答
question = "What are major topics for this class?"
docs = vectordb.similarity_search(question,k=3)
len(docs)
```

```python
# 【真实 Bug 修复】ChatOpenAI：对话式 LLM，用来根据检索到的上下文生成最终回答。
# 原代码 `from langchain_community.chat_models import ChatOpenAI` 在当前 langchain_community 0.4.2 下
# 会 ImportError（ChatOpenAI 已经从 langchain_community 移除，官方推荐统一放进独立包 langchain_openai）。
# temperature=0 表示尽量减少随机性，让回答更确定、可复现，适合问答场景
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model_name=llm_name, temperature=0)
```

```python
# 【真实 Bug 修复】RetrievalQA：把"检索"和"LLM 生成回答"串成一条链的核心组件。
# 原代码 `from langchain_community.chains import RetrievalQA` 会 ImportError
# （langchain_community.chains 这个子模块里根本没有 RetrievalQA）。
# 新版 langchain 1.4.0 下，这类经典 Chain 被归入 langchain_classic 这个兼容层包，正确路径如下。
from langchain_classic.chains import RetrievalQA
```

```python
# from_chain_type 默认用 "stuff" 策略：把检索到的所有文档直接"塞"（stuff）进一个 prompt 里一次性喂给 LLM，
# 实现简单、效果通常不错，但文档太多/太长时容易超出上下文长度限制
qa_chain = RetrievalQA.from_chain_type(
    llm,
    retriever=vectordb.as_retriever()
)
```

```python
# 【版本提示】qa_chain(...) 这种直接调用 Chain 对象的写法（对应 Chain.__call__）在新版 langchain-core 里
# 已被标记为 deprecated（推荐用 qa_chain.invoke(...) 代替），但目前仍然可以正常运行，只是会打印一条 deprecation warning，
# 属于"能跑但不推荐"的情况，这里保留课程原写法不做修改
result = qa_chain({"query": question})
```

```python
# RetrievalQA 返回的是一个 dict，"result" 键对应 LLM 生成的最终回答文本
result["result"]
```

```python
# 自定义 prompt 模板：Prompt Engineering 是 RAG 效果的关键一环。
# {context} 会被自动替换成检索到的文档内容，{question} 替换成用户问题。
# 模板里明确要求"不知道就说不知道，别编答案"（减少幻觉）、限制回答长度（三句话以内），
# 并要求固定结尾语（"thanks for asking!"，方便验证 prompt 真的生效了）
from langchain_core.prompts import PromptTemplate

# Build prompt
template = """Use the following pieces of context to answer the question at the end. If you don't know the answer, just say that you don't know, don't try to make up an answer. Use three sentences maximum. Keep the answer as concise as possible. Always say "thanks for asking!" at the end of the answer.
{context}
Question: {question}
Helpful Answer:"""
QA_CHAIN_PROMPT = PromptTemplate.from_template(template)
```

```python
# chain_type_kwargs={"prompt": QA_CHAIN_PROMPT} 把自定义 prompt 传给 stuff 链
# return_source_documents=True 让返回结果里附带用到的原始文档，方便做"引用来源"展示、核查答案是否有依据
# Run chain
qa_chain = RetrievalQA.from_chain_type(
    llm,
    retriever=vectordb.as_retriever(),
    return_source_documents=True,
    chain_type_kwargs={"prompt": QA_CHAIN_PROMPT}
)
```

```python
question = "Is probability a class topic?"
```

```python
result = qa_chain({"query": question})
```

```python
# 检查回答是否遵守了自定义 prompt 的要求（应该以 "thanks for asking!" 结尾）
result["result"]
```

```python
# 因为传了 return_source_documents=True，可以看到回答具体是依据哪个 Document 生成的，便于溯源核查
result["source_documents"][0]
```

```python
# chain_type="map_reduce"：先对每个检索到的文档块单独让 LLM 生成一个小结（map），
# 再把所有小结汇总起来让 LLM 生成最终答案（reduce）。
# 好处是能处理比 "stuff" 更多/更长的文档（不会一次性超出上下文长度），
# 代价是要多次调用 LLM，速度更慢、成本更高，而且各块之间独立处理，可能丢失跨块的上下文关联
qa_chain_mr = RetrievalQA.from_chain_type(
    llm,
    retriever=vectordb.as_retriever(),
    chain_type="map_reduce"
)
```

```python
result = qa_chain_mr({"query": question})
```

```python
# map_reduce 因为丢失了跨文档块的上下文，回答质量有时反而不如简单的 stuff 策略，可以和上面的结果对比感受一下
result["result"]
```

```python
# 重新跑一遍 map_reduce（如果上面打开了 LangSmith 追踪，这里就能在 LangSmith 后台看到 map/reduce 每一步的详细过程）
qa_chain_mr = RetrievalQA.from_chain_type(
    llm,
    retriever=vectordb.as_retriever(),
    chain_type="map_reduce"
)
result = qa_chain_mr({"query": question})
result["result"]
```

```python
# chain_type="refine"：迭代式策略——先用第一个文档块生成一个初始答案，
# 然后依次用后面的文档块去"精炼/修正"这个答案（每次都把上一轮答案 + 新文档块一起喂给 LLM）。
# 相比 map_reduce 能保留更多跨块的上下文关联，但因为是串行处理，速度通常最慢
qa_chain_mr = RetrievalQA.from_chain_type(
    llm,
    retriever=vectordb.as_retriever(),
    chain_type="refine"
)
result = qa_chain_mr({"query": question})
result["result"]
```

```python
# 回到最基础的 stuff 策略，演示 RetrievalQA 的一个已知局限：它每次调用都是独立的，不具备对话记忆能力
# （下一格连续问两个有上下文关联的问题就能看出这个问题，06 节会引入带记忆的 ConversationalRetrievalChain 来解决）
qa_chain = RetrievalQA.from_chain_type(
    llm,
    retriever=vectordb.as_retriever()
)
```

```python
question = "Is probability a class topic?"
result = qa_chain({"query": question})
result["result"]
```

```python
# 这里的 "those prerequisites" 指代的是上一个问题里提到的内容，但 RetrievalQA 不记得上一轮对话，
# 检索时只会拿 "why are those prerequesites needed?" 这句话本身去做相似度检索，很可能检索不到真正相关的内容，
# 导致回答答非所问——这正是 06 节要解决的"对话记忆"问题
question = "why are those prerequesites needed?"
result = qa_chain({"query": question})
result["result"]
```

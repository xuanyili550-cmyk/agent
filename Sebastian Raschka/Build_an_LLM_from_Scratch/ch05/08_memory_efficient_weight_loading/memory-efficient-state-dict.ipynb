1. Benchmark utilities
First, let's define some utility code to track VRAM (GPU memory)
Later, we will also introduce a tool to track the main system RAM (CPU memory)
The purpose of these functions will become clear when we apply them later

```python
# gc：Python 垃圾回收模块，用于手动触发内存回收
# time：用于在清理显存后等待一段缓冲时间
# torch：PyTorch 主库，用于操作 GPU 显存统计
import gc
import time
import torch


def start_memory_tracking():
    """Initialize GPU memory tracking."""
    # 重置 CUDA 的“峰值显存”统计计数器，这样后续调用 max_memory_allocated()
    # 得到的就是从此刻开始新产生的显存峰值，而不是历史累计的峰值
    if torch.cuda.is_available():
        torch.cuda.reset_peak_memory_stats()
    else:
        print("This notebook is intended for CUDA GPUs but CUDA is not available.")

def print_memory_usage():
    # 读取自上次 reset_peak_memory_stats() 以来 GPU 显存的峰值占用（字节），
    # 并换算成 GB 方便阅读
    max_gpu_memory = torch.cuda.max_memory_allocated() / (1024 ** 3)  # Convert bytes to GB
    print(f"Maximum GPU memory allocated: {max_gpu_memory:.1f} GB")

def cleanup():
    # 触发 Python 垃圾回收，释放不再被引用的 CPU 对象
    gc.collect()
    # 释放 PyTorch 缓存分配器中已不再使用、但仍占用着的显存块，归还给系统/驱动
    torch.cuda.empty_cache()
    time.sleep(3)  # some buffer time to allow memory to clear
    # 再次重置峰值显存统计，方便下一次实验从 0 开始重新计量
    torch.cuda.reset_peak_memory_stats()
    # 注意：这里用到的 device 是全局变量，在下面构建/搬运模型的单元格中才会被赋值；
    # 只要 cleanup() 在那之后调用就不会报 NameError
    max_memory_allocated = torch.cuda.max_memory_allocated(device) / (1024 ** 3)
    print(f"Maximum GPU memory allocated: {max_memory_allocated:.1f} GB")
```

3. Basic weight loading

```python
# Then load pretrained weights

start_memory_tracking()

# 【做法一：最朴素的加载方式】
# 先在 CPU 上创建一份完整的（随机初始化的）模型参数
model = GPTModel(BASE_CONFIG)
# 再把这份 CPU 参数整体搬到 GPU 上 —— 此时显存里已经有一份随机初始化的模型权重
model.to(device)

# torch.load(..., map_location=device) 会把磁盘上保存的 state_dict
# 直接反序列化到 GPU 显存中，相当于在显存里又生成了“第二份”权重（训练好的那份）；
# 随后 load_state_dict 把这份权重的数值拷贝进模型现有参数里。
# 问题：整个过程中，GPU 显存里一度会同时存在“模型自身参数”和“待加载的 state_dict”
# 两份完整权重，显存峰值几乎翻倍——这正是本节后续要优化解决的问题
model.load_state_dict(
    torch.load("model.pth", map_location=device, weights_only=True)
)
model.to(device)
model.eval();

print_memory_usage()
```

```python
# Test if the model works (no need to track memory here)
# 简单前向验证：确认加载完权重后的模型依然能够正常推理
test_input = torch.tensor([[1, 2, 3]]).to(device)
model.eval()

with torch.no_grad():
    model(test_input)

# 清理本次实验占用的显存，为下一种加载方式的对比做准备
del model, test_input
cleanup()
```

```python
start_memory_tracking()

# 【做法二：先在 CPU 上完成权重加载，再把合并好的模型整体搬到 GPU】
# 注意这里没有像上一小节那样先 model.to(device)，
# 模型参数此时仍然留在 CPU 内存中
model = GPTModel(BASE_CONFIG)

# map_location="cpu"：把磁盘上的 state_dict 反序列化到 CPU 内存（而不是 GPU 显存），
# 这样“模型参数”和“待加载权重”这两份数据此时都在 CPU 上，
# 不会像上一小节那样在 GPU 显存里同时出现两份权重叠加导致的峰值
model.load_state_dict(
    torch.load("model.pth", map_location="cpu", weights_only=True)
)
# 加载合并完成、CPU 上临时的 state_dict 引用被覆盖释放后，
# 再把合并好的模型整体搬到 GPU，这样 GPU 显存中只需要保留一份最终权重
model.to(device)
model.eval();

print_memory_usage()
```

```python
# Test if the model works (no need to track memory here)
# 简单前向验证：确认加载完权重后的模型依然能够正常推理
test_input = torch.tensor([[1, 2, 3]]).to(device)
model.eval()

with torch.no_grad():
    model(test_input)

# 清理本次实验占用的显存，为下一种加载方式的对比做准备
del model, test_input
cleanup()
```

4. Loading weights sequentially

```python
start_memory_tracking()

# 【做法三：逐张量（sequential）拷贝权重】
# 先在 GPU 上创建好模型结构和随机初始化的参数
model = GPTModel(BASE_CONFIG).to(device)

# 把整份 state_dict 加载到 CPU 内存中（而不是一次性搬到 GPU），
# 这样 GPU 显存里此时只多出“模型自身”这一份参数，没有额外的整份权重副本
state_dict = torch.load("model.pth", map_location="cpu", weights_only=True)

print_memory_usage()

# Sequentially copy weights to the model's parameters
# 核心思路：不是把整个 state_dict 一次性搬到 GPU 再整体拷贝，
# 而是逐个参数张量地进行拷贝——每次只把“当前这一个”张量从 CPU 搬到 GPU 并写入模型，
# 这样任意时刻 GPU 上“额外”占用的显存只有当前这一个张量的大小，
# 而不是整份权重的大小，从而显著降低显存峰值
with torch.no_grad():
    for name, param in model.named_parameters():
        if name in state_dict:
            # .to(device) 把当前这一个张量搬到 GPU 上，
            # copy_ 原地把数值写入模型对应的参数（不改变参数的内存地址/梯度追踪状态）
            param.copy_(state_dict[name].to(device))
        else:
            print(f"Warning: {name} not found in state_dict.")

print_memory_usage()
```

```python
# Test if the model works (no need to track memory here)
test_input = torch.tensor([[1, 2, 3]]).to(device)
model.eval()

with torch.no_grad():
    model(test_input)

# 注意：这里的 param 是上一个单元格里 for 循环结束后残留下来的循环变量
# （Python 的 for 循环变量在循环结束后依然存在于当前作用域中），
# 一并 del 掉是为了确保它所引用的那个张量也能被正确释放
del model, test_input, state_dict, param
cleanup()
```

5. Loading the model with low CPU memory

```python
# os：用于获取当前进程 PID
# psutil：跨平台的系统/进程资源监控库，这里用来读取进程的常驻内存（RSS）
# Thread：用后台线程持续采样内存占用，这样无需修改被测函数本身
import os
import psutil
from threading import Thread


def memory_usage_in_gb(func, *args, **kwargs):
    # 获取当前 Python 进程对象，用于查询其内存占用情况
    process = psutil.Process(os.getpid())

    # Measure the baseline memory usage before running the function
    # RSS（Resident Set Size）：进程实际占用的物理内存，这里近似代表 CPU 内存占用
    baseline_mem = process.memory_info().rss / 1024 ** 3  # in GB

    # Start monitoring memory in a separate thread
    mem_usage = []
    done = False

    def monitor_memory():
        # 每隔 0.1 秒采样一次当前进程的 RSS 内存占用；
        # 因为内存峰值可能出现在 func 执行过程中的某个瞬间，
        # 只在 func 执行前后各测一次是抓不到真实峰值的
        while not done:
            mem_usage.append(process.memory_info().rss / 1024 ** 3)  # Convert to GB
            time.sleep(0.1)

    # 启动后台采样线程，与主线程并行运行
    t = Thread(target=monitor_memory)
    t.start()

    # Run the function
    # 在主线程里正常执行被测函数（例如某一种权重加载方式）
    func(*args, **kwargs)

    # Stop monitoring
    # 通知采样线程停止，并等待它结束
    done = True
    t.join()

    # 用采样到的内存峰值减去基线内存，得到 func 执行期间“额外”增加的内存峰值
    peak_mem_usage_gb = max(mem_usage) - baseline_mem
    return peak_mem_usage_gb
```

```python
# 把前面“逐张量加载”的流程封装成函数，方便用 memory_usage_in_gb
# 测量它执行期间的 CPU 内存峰值（而不仅仅是 GPU 显存峰值）
def load_sequentially():
    start_memory_tracking()

    model = GPTModel(BASE_CONFIG).to(device)

    # 完整的 state_dict 会先被反序列化到 CPU 内存中，这部分正是 CPU 内存峰值的主要来源
    state_dict = torch.load("model.pth", map_location="cpu", weights_only=True)

    print_memory_usage()

    # Sequentially copy weights to the model's parameters
    with torch.no_grad():
        for name, param in model.named_parameters():
            if name in state_dict:
                param.copy_(state_dict[name].to(device))
            else:
                print(f"Warning: {name} not found in state_dict.")

    print_memory_usage()


# 用后台线程采样的方式，测出执行 load_sequentially() 期间 CPU 内存的峰值增量
peak_memory_used = memory_usage_in_gb(load_sequentially)
print(f"-> Maximum CPU memory allocated: {peak_memory_used:.1f} GB")
```

```python
# 【关键优化：结合 meta device 的逐张量加载】
def load_sequentially_with_meta():
    start_memory_tracking()

    # torch.device("meta") 是一种“虚拟设备”：在这个上下文里创建的张量
    # 只记录形状(shape)、数据类型(dtype)等元信息，并不会真正分配底层存储、
    # 也不占用任何 CPU 或 GPU 内存。这样即使模型有上亿参数，
    # 用 meta device 构造它也几乎是瞬时完成、且零内存开销的
    with torch.device("meta"):
        model = GPTModel(BASE_CONFIG)

    # to_empty(device=...) 把“meta 模型”落地到真实设备上：
    # 按照 meta 张量记录的形状，在目标设备（这里是 GPU）上分配未初始化的存储空间；
    # 注意此时参数的数值是未初始化的垃圾值，还没有被赋予真实权重
    model = model.to_empty(device=device)

    # 这里直接用 map_location=device，把 state_dict 反序列化到 GPU 上，
    # 因为本函数要重点节省的是 CPU 内存（memory_usage_in_gb 测的就是 CPU 内存）
    state_dict = torch.load("model.pth", map_location=device, weights_only=True)

    print_memory_usage()

    # Sequentially copy weights to the model's parameters
    # 权重此时已经在 GPU 上了，不需要再 .to(device)，
    # 逐个张量地把训练好的权重拷贝进（已经在 GPU 上分配好存储的）模型参数中
    with torch.no_grad():
        for name, param in model.named_parameters():
            if name in state_dict:
                param.copy_(state_dict[name])
            else:
                print(f"Warning: {name} not found in state_dict.")

    print_memory_usage()

# 测量：meta device 构造（零 CPU 内存开销）+ 直接把权重加载到 GPU，
# 使整个过程中 CPU 内存几乎不需要额外分配，
# 相比 load_sequentially() 应能观察到明显更低的 CPU 内存峰值
peak_memory_used = memory_usage_in_gb(load_sequentially_with_meta)
print(f"-> Maximum CPU memory allocated: {peak_memory_used:.1f} GB")
```

```python
# 把最开始“最朴素的加载方式”也封装成函数，作为 CPU 内存维度的基线（baseline）对比对象
def baseline():
    start_memory_tracking()

    # 先在 CPU 上创建模型（会真实占用一份 CPU 内存）
    model = GPTModel(BASE_CONFIG)
    model.to(device)

    # torch.load(..., map_location=device) 把整份 state_dict 直接反序列化到 GPU，
    # 这一步本身不太占 CPU 内存；但前面 GPTModel(BASE_CONFIG) 在 CPU 上构造模型
    # 已经产生了一份完整的 CPU 内存占用，可作为与 meta device 方案对比的基线
    model.load_state_dict(torch.load("model.pth", map_location=device, weights_only=True))
    model.to(device)
    model.eval();

    print_memory_usage()

# 对比：baseline（普通构造）vs load_sequentially（逐张量 + CPU 中转加载）
# vs load_sequentially_with_meta（meta device + 逐张量），
# 预期能看到 CPU 内存峰值依次明显下降
peak_memory_used = memory_usage_in_gb(baseline)
print(f"-> Maximum CPU memory allocated: {peak_memory_used:.1f} GB")
```

6. Using mmap=True (recommmended)

```python
# 【推荐做法：meta device + mmap=True + assign=True】
def best_practices():
  # 同样先用 meta device 构造模型骨架，不占用真实内存/显存
  with torch.device("meta"):
      model = GPTModel(BASE_CONFIG)

  # mmap=True：以内存映射（memory-mapped file）的方式读取权重文件，
  #   不会把整个文件一次性读入内存，而是让操作系统按需分页加载，
  #   可以大幅降低加载大模型权重时的 CPU 内存占用
  # assign=True：不再对模型中已存在的参数张量做原地 copy_ 拷贝数值，
  #   而是直接用 state_dict 中的张量“替换”模型的参数（相当于赋值而不是拷贝），
  #   从而省去为 meta 模型的参数额外分配一次真实存储、再把数据拷贝进去的过程，
  #   进一步降低内存/显存的中间峰值
  model.load_state_dict(
      torch.load("model.pth", map_location=device, weights_only=True, mmap=True),
      assign=True
  )

  print_memory_usage()

# 这是本节推荐的最佳实践组合：meta device（零成本构造）
# + mmap（按需读取权重文件）+ assign=True（避免额外拷贝），
# 通常能在本 notebook 的几种方式中拿到最低的 CPU 内存峰值
peak_memory_used = memory_usage_in_gb(best_practices)
print(f"-> Maximum CPU memory allocated: {peak_memory_used:.1f} GB")
```

7. Other methods

```python
# 【另一种思路：把权重拆分成多个独立文件，分别保存/加载】
model = GPTModel(BASE_CONFIG)
# Assume `model` is your trained model
state_dict = model.state_dict()

# Create a directory to store individual parameter files
os.makedirs("model_parameters", exist_ok=True)

# Save each parameter tensor separately
# 不再把权重保存成一个大的 state_dict 文件，而是每个参数张量单独存成一个 .pt 文件，
# 这样后面加载时就可以做到真正的“按需、逐个”读取，
# 且任意时刻磁盘 IO 涉及的都只是单个张量的大小
for name, param in state_dict.items():
    torch.save(param.cpu(), f"model_parameters/{name}.pt")

del model
```

```python
def load_individual_weights():

    start_memory_tracking()

    # 依旧用 meta device 零成本构造模型骨架
    with torch.device("meta"):
        model = GPTModel(BASE_CONFIG)

    # 在目标设备（GPU）上为参数分配真实存储空间（此时数值仍是未初始化的垃圾值）
    model = model.to_empty(device=device)

    print_memory_usage()
    param_dir = "model_parameters"

    with torch.no_grad():
        for name, param in model.named_parameters():
            weight_path = os.path.join(param_dir, f"{name}.pt")
            if os.path.exists(weight_path):
                # 每次只从磁盘读取“当前这一个参数”对应的独立文件，
                # 而不是像前面那样一次性把整份 state_dict 读入内存
                param_data = torch.load(weight_path, map_location="cpu", weights_only=True)
                # copy_ 会自动处理 CPU -> GPU 的数据搬运，把数值写入模型参数
                param.copy_(param_data)
                del param_data  # Free memory
            else:
                print(f"Warning: {name} not found in {param_dir}.")

    print_memory_usage()


# 由于任意时刻内存中只需要驻留“一个参数张量”的临时副本，
# 这种方式通常能把 CPU 内存峰值压到最低，是逐张量加载思路的极致体现
peak_memory_used = memory_usage_in_gb(load_individual_weights)
print(f"-> Maximum CPU memory allocated: {peak_memory_used:.1f} GB")
```

```python
import torch

# 定义一个简单的前馈神经网络(MLP),用于练习 A.4:统计模型的可训练参数量
# 该网络包含 2 个隐藏层和 1 个输出层,均为全连接(Linear)+ ReLU 激活的结构
class NeuralNetwork(torch.nn.Module):
    def __init__(self, num_inputs, num_outputs):
        # num_inputs: 输入特征维度
        # num_outputs: 输出维度(如分类任务的类别数)
        super().__init__()

        self.layers = torch.nn.Sequential(

            # 1st hidden layer
            # 第 1 个隐藏层:Linear(num_inputs -> 30)
            # 权重形状 (30, num_inputs),偏置形状 (30,)
            torch.nn.Linear(num_inputs, 30),
            torch.nn.ReLU(),

            # 2nd hidden layer
            # 第 2 个隐藏层:Linear(30 -> 20)
            # 权重形状 (20, 30),偏置形状 (20,)
            torch.nn.Linear(30, 20),
            torch.nn.ReLU(),

            # output layer
            # 输出层:Linear(20 -> num_outputs),不接激活函数(直接返回 logits)
            # 权重形状 (num_outputs, 20),偏置形状 (num_outputs,)
            torch.nn.Linear(20, num_outputs),
        )

    def forward(self, x):
        # x 形状: (batch_size, num_inputs)
        # 经过 self.layers 前向传播后得到 logits,形状: (batch_size, num_outputs)
        logits = self.layers(x)
        return logits
# 实例化模型:输入维度为 2,输出维度为 2(与本章正文中的示例模型结构一致)
model = NeuralNetwork(2, 2)

# 统计模型中所有 requires_grad=True 的参数张量的元素总数之和,即可训练参数总量
# p.numel() 返回该参数张量中元素个数(例如某权重矩阵形状为 (out, in),则贡献 out*in 个参数)
num_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print("Total number of trainable model parameters:", num_params)
```

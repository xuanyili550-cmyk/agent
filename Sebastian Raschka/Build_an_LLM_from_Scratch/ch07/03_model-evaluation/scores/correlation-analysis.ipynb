Score Correlation Analysis

```
import json

# 读入两位「裁判」(GPT-4 与 Llama3-8B) 对同一模型(model-1)各条回复给出的分数列表，
# 后面用来分析两位裁判打分的一致性/相关性
with open("gpt4-model-1-response.json", "r") as file:
    gpt4_model_1 = json.load(file)

with open("llama3-8b-model-1-response.json", "r") as file:
    llama3_8b_model_1 = json.load(file)
```

```
import numpy as np
import matplotlib.pyplot as plt


list1, list2 = gpt4_model_1, llama3_8b_model_1  # x=GPT-4 分数, y=Llama3-8B 分数

plt.scatter(list1, list2)  # 散点图直观看两位裁判打分的关系
# 用一次多项式(线性)拟合并画出趋势线：polyfit 求斜率/截距，poly1d 构造函数，在去重后的 x 上求值
plt.plot(
    np.unique(list1),
    np.poly1d(np.polyfit(list1, list2, 1))(np.unique(list1))
)
plt.xlabel("GPT-4")
plt.ylabel("Llama3 8B")
plt.show()
```

```
import pandas as pd
from scipy.stats import spearmanr, kendalltau

# 三种相关系数衡量两位裁判评分的一致性：
pearson_correlation = np.corrcoef(list1, list2)[0, 1]      # 皮尔逊：线性相关强度
spearman_correlation, _ = spearmanr(list1, list2)          # 斯皮尔曼：基于秩次的单调相关
kendall_tau_correlation, _ = kendalltau(list1, list2)      # 肯德尔 Tau：基于「同序/逆序对」的秩相关

# 汇总成一行表格，便于展示
correlation_table = pd.DataFrame({
    "Pearson": [pearson_correlation],
    "Spearman": [spearman_correlation],
    "Kendall Tau": [kendall_tau_correlation]
}, index=['Results'])

correlation_table
```
